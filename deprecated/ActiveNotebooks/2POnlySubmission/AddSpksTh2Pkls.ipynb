```python
def load_single_sess(mouse, deets, verbose = True, pkl_basedir = '/home/mplitt/YMazeSessPkls'):
    pkldir = os.path.join(pkl_basedir, mouse)            
    
    sess = stx.session.YMazeSession.from_file(
            os.path.join(pkldir, deets['date'], "%s_%d.pkl" % (deets['scene'], deets['session'])),
            verbose=False, novel_arm=deets['novel_arm'])
    return sess


def thresh_spks(sess):

    dff = np.copy(sess.timeseries['F_dff'])
    mad = sp.stats.median_abs_deviation(dff, axis=-1, nan_policy='omit')[:,np.newaxis]

    dff -= 4*mad
    dff[dff<0]=0
    dff_bin = dff>0

    spks = np.copy(sess.timeseries['spks'])
    spks[~dff_bin]=0
    spks_bin = spks>0
    
    # spks_bin = 1.*(np.diff(spks_bin,axis=1,prepend=0)>0)
    sess.add_timeseries(spks_th = spks)
    sess.add_pos_binned_trial_matrix('spks_th')

    sess.add_timeseries(F_dff_th = dff)
    sess.add_pos_binned_trial_matrix('F_dff_th')

    sess.add_timeseries(F_dff_bin = 1.*(dff>0))
    sess.add_pos_binned_trial_matrix('F_dff_bin')
        
    speed = np.nan*sess.vr_data['t'].to_numpy()
    block = np.nan*sess.vr_data['t'].to_numpy()
    for t, (start,stop) in enumerate(zip(sess.trial_start_inds.to_list(), sess.teleport_inds.to_list())):
        speed[start:stop] = 10*sess.scan_info['frame_rate']*np.diff(sess.vr_data['t'].iloc[start-1:stop])
        block[start:stop] = sess.trial_info['block_number'][t]
        
    sess.add_timeseries(speed=speed)
    sess.add_pos_binned_trial_matrix('speed')
    
    spks_ = np.copy(sess.timeseries['spks'])
    spks_[:,speed<2]=np.nan
    sess.add_timeseries(spks_nostop=spks)
    sess.add_pos_binned_trial_matrix('spks_nostop')

    dff_ = np.copy(sess.timeseries['F_dff'])
    dff_[:,speed<2]=np.nan
    sess.add_timeseries(F_dff_nostop=dff_)
    sess.add_pos_binned_trial_matrix('F_dff_nostop')
    
    sess.add_timeseries(LR = sess.vr_data['LR'].to_numpy())
    sess.add_timeseries(t = sess.vr_data['t'].to_numpy())
    sess.add_timeseries(reward = sess.vr_data['reward'].to_numpy())
    sess.add_timeseries(block_number = block)
    return sess
```

```python
for session_dict in (stx.ymaze_sess_deets.CTRL_sessions, stx.ymaze_sess_deets.KO_sessions):
# for session_dict in (stx.ymaze_sess_deets.KO_sessions,):
    for mouse, sessions in session_dict.items():
        print(mouse)
        for deets in sessions:
            if isinstance(deets,tuple):
                for _deets in deets:
                    print(_deets)
                    sess = load_single_sess(mouse,_deets)
                    sess = thresh_spks(sess)
                    # sess.place_cells_calc(Fkey='spks_th', nperms=1000)
                    sess._abc_impl = None
                    tpu.sess.save_session(sess,'/home/mplitt/YMazeSessPkls')
                    # upsample_tr/ial_mats(sess)
            else:
                print(deets)
                sess = load_single_sess(mouse,deets)
                sess = thresh_spks(sess)
                # sess.place_cells_calc(Fkey='spks_th', nperms=1000)
                sess._abc_impl = None
                tpu.sess.save_session(sess,'/home/mplitt/YMazeSessPkls')
```

```python
def run_dcnv(sess):
    for chan in ('channel_0', 'channel_1'):
        tau = sess.s2p_ops[chan]['tau']
        start_inds = sess.trial_starts[chan]
        end_inds = sess.trial_ends[chan]
        dff = sess.timeseries[f'{chan}_F_dff']
        spks = np.nan*np.zeros_like(dff)
        for block in np.unique(sess.trial_info['block_number']).tolist():
            start_ind = start_inds[sess.trial_info['block_number'] == block][0]
            stop_ind = end_inds[sess.trial_info['block_number'] == block][-1]
            print(start_ind, stop_ind)

            curr_idx = np.arange(start_ind, stop_ind)
            
            spks[:, curr_idx] = stx.session.dcnv.oasis(dff[:, curr_idx], 2000, tau, sess.scan_info['frame_rate'])
        sess.timeseries[f'{chan}_spks']=spks
        sess.add_pos_binned_trial_matrix_mux(f'{chan}_spks',channel=chan)
   

def thresh_spks_sparse(sess):


    speed = np.nan*sess.vr_data['t'].to_numpy()
    for t, (start,stop) in enumerate(zip(sess.trial_start_inds.to_list(), sess.teleport_inds.to_list())):
        speed[start:stop] = 10*sess.scan_info['frame_rate']*np.diff(sess.vr_data['t'].iloc[start-1:stop])
        
    sess.add_timeseries(speed=speed)
    sess.add_pos_binned_trial_matrix('speed')
    run_dcnv(sess)
    # sess.add_timeseries(LR = sess.vr_data['LR'].to_numpy())
    # sess.add_timeseries(t = sess.vr_data['t'].to_numpy())
    # sess.add_timeseries(reward = sess.vr_data['reward'].to_numpy())
    # sess.add_timeseries(block_number = block
    # sess.neuropil_corrected_dff_mux(channels=('channel_0', 'channel_1'))

    for chan in ('channel_0', 'channel_1'):

        
        
        dff = np.copy(sess.timeseries[f'{chan}_F_dff'])
        mad = sp.stats.median_abs_deviation(dff, axis=-1, nan_policy='omit')[:,np.newaxis]

        dff -= 4*mad
        dff[dff<0]=0
        dff_bin = dff>0

        
        spks = np.copy(sess.timeseries[f'{chan}_spks'])
        spks[~dff_bin]=0
        spks_bin = spks>0

        # spks_bin = 1.*(np.diff(spks_bin,axis=1,prepend=0)>0)

        sess.timeseries[f'{chan}_spks_th'] = spks
        sess.add_pos_binned_trial_matrix_mux(f'{chan}_spks_th', channel=chan)

        sess.timeseries[f'{chan}_F_dff_th'] = dff
        sess.add_pos_binned_trial_matrix_mux(f'{chan}_F_dff_th', channel=chan)

        sess.timeseries[f'{chan}_F_dff_bin'] = 1.*(dff>0)
        sess.add_pos_binned_trial_matrix_mux(f'{chan}_F_dff_bin',channel=chan)
        
    return sess

```
