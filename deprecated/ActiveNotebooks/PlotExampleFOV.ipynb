```python
import os

import numpy as np
import scipy as sp
from matplotlib import pyplot as plt
from matplotlib import gridspec

import TwoPUtils as tpu
import STX3KO_analyses as stx
from STX3KO_analyses import utilities as u

ko_mice = stx.ymaze_sess_deets.ko_mice
ctrl_mice = stx.ymaze_sess_deets.ctrl_mice


%matplotlib inline

%load_ext autoreload
%autoreload 2
```

```python
figdir = "/mnt/BigDisk/YMazeResults/neural/example_FOVs"
os.makedirs(figdir, exist_ok=True)
```

```python
sess = u.single_mouse_concat_sessions(ko_mice[3], date_inds = np.arange(8), load_ops = True)
```

```python
len(sess.s2p_ops)
```

```python
for i, ops in enumerate(sess.s2p_ops):
    fig, ax = plt.subplots(3,1,figsize=[5,15])
    cimg = np.zeros([100,150,3])
    green = ops['meanImg'][225:325,300:450]
    green /= np.amax(green.ravel())
    red = ops['meanImg_chan2'][225:325,300:450]
    red /= np.amax(red.ravel())
    cimg[:,:,1] = np.clip(green, 0 ,.6)/.6
    cimg[:,:,0] = red
    cimg[:,:,2] = red 
    
    ax[0].imshow(green, cmap = "Greys_r", aspect = 796/512.)
    ax[1].imshow(red, cmap = "Greys_r", aspect = 796/512.)
    
    
    ax[2].imshow(cimg, aspect = 796/512.)
    
    fig.subplots_adjust(hspace=.1)
    
    fig.savefig(os.path.join(figdir,"%s_%d.pdf" % (ko_mice[3], i)))
```
