```python
import os
import pickle
import pathlib

import numpy as np
import scipy as sp
from matplotlib import pyplot as plt
from matplotlib import gridspec
import seaborn as sns
import pandas as pd
import pingouin as pg

import STX3KO_analyses as stx
from STX3KO_analyses import utilities as u
import TwoPUtils as tpu

ko_mice = stx.ymaze_sess_deets.ko_mice
ctrl_mice = stx.ymaze_sess_deets.ctrl_mice
sparse_mice = stx.ymaze_sess_deets.sparse_mice


plt.rcParams['pdf.fonttype']=42
%matplotlib inline

%load_ext autoreload
%autoreload 2
```

```python
figdir = pathlib.Path('/mnt/BigDisk/fig_scratch/stx3/figS5_pc_fracs')
figdir.mkdir(exist_ok=True, parents=True)
```

```python

# fraction of cells that are place cells dense
rows = []
for cond, mice in zip(('ctrl','ko'), (ctrl_mice, ko_mice)):
    for mouse in mice:
        for day in range(6):

            sess = u.load_single_day(mouse,day)

            for ttype in ('fam','nov'):
                if ttype=='fam':
                    mask = sess.fam_place_cell_mask()
                else:
                    mask = sess.nov_place_cell_mask()

                pc_frac = float(mask.sum())/mask.shape[0]

                rows.append({
                    'cond': cond,
                    'mouse': mouse,
                    'day': day,
                    'nov': ttype,
                    'frac': pc_frac,
                })
df = pd.DataFrame(rows)
```

```python
fig, ax = plt.subplots(1,2, figsize=[9,3])

for i, ttype in enumerate(('fam', 'nov')):
    tmp_df = df.loc[df['nov']==ttype]
    if ttype == 'fam':
        marker='.'
        s=12
    else:
        marker='P'
        s=8

    sns.stripplot(
        data=tmp_df,
        x='day',
        y='frac',
        hue='cond',
        dodge=True,
        ax=ax[i],
        palette=['black','red'],
        hue_order=['ctrl','ko'],
        alpha=1.,
        marker = marker,
        s=s,
    )

    sns.pointplot(
        data=tmp_df,
        x='day',
        y='frac',
        hue='cond',
        dodge=.4,
        ax=ax[i],
        palette=['black','red'],
        hue_order=['ctrl','ko'],
        errorbar=None,
        linestyles="none",
        marker="_", markersize=15, markeredgewidth=5, legend=False, alpha=.4,
    )
    ax[i].set_title(ttype)


for a in ax.flatten():
    a.spines['top'].set_visible(False)
    a.spines['right'].set_visible(False)

fig.savefig(figdir / "dense_frac_pcs.pdf")
```

```python
rows = []
for mouse in sparse_mice:
    for day in range(6):
        if mouse == 'SparseKO_09' and day == 2:
            continue
        sess = u.load_single_day(mouse,day)


        for chan in ('channel_0', 'channel_1'):
            for ttype in ('fam', 'nov'):
                if ttype == 'fam':
                    mask = sess.fam_place_cell_mask(chan=chan, mux=True)
                else:
                    mask = sess.nov_place_cell_mask(chan=chan, mux=True)

                pc_frac = float(mask.sum())/mask.shape[0]
                
                rows.append({
                    'mouse': mouse,
                    'day': day,
                    'channel': chan,
                    'ttype': ttype,
                    'frac': pc_frac,
                })

df = pd.DataFrame(rows)
```

```python
fig, ax = plt.subplots(1,2, figsize=[9,3])

for i, ttype in enumerate(('fam', 'nov')):
    tmp_df = df.loc[df['ttype']==ttype]
    if ttype == 'fam':
        marker='.'
        s=12
    else:
        marker='P'
        s=8
    sns.stripplot(
        data=tmp_df,
        x='day',
        y='frac',
        hue='channel',
        dodge=True,
        ax=ax[i],
        palette=['black','red'],
        hue_order=['channel_1','channel_0'],
        alpha=1., legend=False, marker=marker, s=s,
    )

    
    ax[i].set_title(ttype)
    for d in range(0,12,2):
        for (x0, y0), (x1, y1) in zip(ax[i].collections[d].get_offsets(), ax[i].collections[d+1].get_offsets()):
            ax[i].plot([x0, x1], [y0, y1], color='black', ls=':', zorder=0)

for a in ax.flatten():
    a.spines['top'].set_visible(False)
    a.spines['right'].set_visible(False)

fig.savefig(figdir / "sparse_frac_pcs.pdf")
```
