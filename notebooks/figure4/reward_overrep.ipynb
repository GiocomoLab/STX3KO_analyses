```python
import pathlib
import numpy as np
import scipy as sp
from matplotlib import pyplot as plt
import pandas as pd
import seaborn as sns


import STX3KO_analyses as stx
from STX3KO_analyses import utilities as u

ko_mice = stx.ymaze_sess_deets.ko_mice
ctrl_mice = stx.ymaze_sess_deets.ctrl_mice

plt.rcParams['pdf.fonttype']=42

%matplotlib inline

%load_ext autoreload
%autoreload 2
```

```python
# proportion of place fields in the reward zone
```

```python
frac_cls = stx.reward_overrep.PeriRewardCellFrac_Dense(
    {'ctrl': ctrl_mice, 'ko': ko_mice},
    np.arange(6), place_cell_only=True, ts_key='F_dff')
```

```python

fig,ax = plt.subplots(1,2, figsize=[9,3], sharey = True)
for i, nov in enumerate(('fam','nov')):
    sns.stripplot(data=frac_cls.df.loc[frac_cls.df['ttype']==nov,:],
                x='day',
                y='frac',
                hue='cond', 
                hue_order=('ko', 'ctrl'), 
                palette=['red', 'black'],
                dodge=True,
                ax=ax[i],
                legend=False)
    


    sns.pointplot(data=frac_cls.df.loc[frac_cls.df['ttype']==nov,:],
              x='day',
              y='frac',
              errorbar=None,
              hue='cond',
              hue_order = ['ko', 'ctrl'],
              dodge = .4,
              palette = ['red','black'],
              ax=ax[i],
              linestyles="none",
              marker="_", markersize=15, markeredgewidth=5, legend=False, alpha=.4,
              )
    ax[i].set_title(nov)

for a in ax.flatten():
    a.spines['top'].set_visible(False)
    a.spines['right'].set_visible(False)
    a.hlines(5./30, -.5,5.5, linestyle='--', color='black')

fig.savefig( figdir / 'frac_per_day.pdf', bbox_inches='tight')

```

```python
fig, ax = plt.subplots(figsize=[3,3])
sns.scatterplot(data=frac_cls.df,
                x='licks',
                y='frac',
                hue='cond',
                style='ttype',
                markers={'fam':'o', 'nov':'P'},
                alpha=.4,
                palette = {'ko':'red', 'ctrl':'black'},
                ax=ax, 
                )
sns.move_legend(ax, "upper left", bbox_to_anchor=(1,1))
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_xlim([0,1.5])
fig.savefig(figdir / 'licks_vs_reward_frac.pdf', bbox_inches='tight')


r,p = sp.stats.spearmanr(frac_cls.df['licks'], frac_cls.df['frac'])
print(f'r={r:.2f}, p={p:.3f}')
```

```python
fig, ax = plt.subplots()
sns.scatterplot(data=frac_cls.df,
                x='speed',
                y='frac',
                hue='cond',
                style='ttype',
                markers={'fam':'o', 'nov':'P'},
                palette = {'ko':'red', 'ctrl':'black'},
                ax=ax, 
                )


r,p = sp.stats.spearmanr(frac_cls.df['speed'], frac_cls.df['frac'])
print(f'r={r:.2f}, p={p:.3f}')
```

```python
fig,ax = plt.subplots( figsize=[4.5,3], sharey = True)
df_mu = frac_cls.df.groupby(['mouse','day','cond'])['frac'].mean().reset_index()
sns.stripplot(data=frac_cls.df.loc[frac_cls.df['ttype']=='fam',:],
            x='day',
            y='frac',
            hue='cond', 
            hue_order=('ko', 'ctrl'), 
            palette=['red', 'black'],
            marker='o',
            alpha=.4,
            s=5,
            dodge=True,
            ax=ax,
            legend=False)

sns.stripplot(data=frac_cls.df.loc[frac_cls.df['ttype']=='nov',:],
            x='day',
            y='frac',
            hue='cond', 
            hue_order=('ko', 'ctrl'), 
            palette=['red', 'black'],
            marker='P',
            alpha=.4,
            s=5,
            dodge=True,
            ax=ax,
            legend=False)



sns.pointplot(data=df_mu,
            x='day',
            y='frac',
            errorbar=None,
            hue='cond',
            hue_order = ['ko', 'ctrl'],
            dodge = .4,
            palette = ['red','black'],
            ax=ax,
            linestyles="none",
            marker="_", markersize=15, markeredgewidth=5, legend=False, alpha=1.,
            )
ax.set_title('combined')


ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.hlines(5./30, -.5,5.5, linestyle='--', color='black')

fig.savefig( figdir / 'frac_per_day_combined.pdf', bbox_inches='tight')
```
