```python
# permutation test
import itertools

rank_sum_shuffle = []
rng = np.random.default_rng()
all_mice = ctrl_mice+ko_mice


for i, p in enumerate(itertools.combinations(ctrl_mice+ko_mice, len(ctrl_mice))):
    df_shuff = frac_cls.df.copy()
    df_shuff['cond']='ko'
    regexp = '|'.join(p)
    mask = df_shuff['mouse'].str.contains(regexp)
    df_shuff.loc[mask,'cond']='ctrl'

    stat, p = sp.stats.ranksums(df_shuff.loc[df_shuff['cond']=='ctrl','frac'],df_shuff.loc[df_shuff['cond']=='ko','frac'])
    rank_sum_shuffle.append(stat)
    if i%1000==0:
        print(i)
```

```python
true_rank_sum, _ = sp.stats.ranksums(frac_cls.df.loc[frac_cls.df['cond']=='ctrl','frac'],
                                     frac_cls.df.loc[frac_cls.df['cond']=='ko','frac'])
```

```python
fig, ax = plt.subplots()
ax.hist(rank_sum_shuffle, 100,density=True)
ax.vlines(true_rank_sum,0,.4, color='red')
```

```python
float((true_rank_sum<np.array(rank_sum_shuffle)).sum()-1)/len(rank_sum_shuffle)
```
