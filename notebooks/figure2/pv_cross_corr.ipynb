```python
import pathlib
import numpy as np
import scipy as sp
from matplotlib import pyplot as plt
import pandas as pd

import STX3KO_analyses as stx
from STX3KO_analyses import utilities as u

ko_mice = stx.ymaze_sess_deets.ko_mice
ctrl_mice = stx.ymaze_sess_deets.ctrl_mice
sparse_mice = stx.ymaze_sess_deets.sparse_mice

plt.rcParams['pdf.fonttype']=42
plt.rcParams['font.family'] ='sans-serif'
plt.rcParams['font.serif'] = 'Arial'


%matplotlib inline

%load_ext autoreload
%autoreload 2
```

```python
# pv correlation across days 
tmat_key = 'F_dff'
ctrl_corr = {'fam': [], 'nov': []}
for mouse in ctrl_mice:
    pv_corr = stx.spatial_analyses.run_acrossday_corr(mouse, fam=True, key=tmat_key, method='cell')
    ctrl_corr['fam'].append(pv_corr)

    pv_corr = stx.spatial_analyses.run_acrossday_corr(mouse, fam=False, key=tmat_key, method='cell')
    ctrl_corr['nov'].append(pv_corr)


ko_corr = {'fam': [], 'nov':[]}
for mouse in ko_mice:
    pv_corr = stx.spatial_analyses.run_acrossday_corr(mouse, fam=True, key=tmat_key, method='cell')
    ko_corr['fam'].append(pv_corr)

    pv_corr = stx.spatial_analyses.run_acrossday_corr(mouse, fam=False, key=tmat_key, method='cell')
    ko_corr['nov'].append(pv_corr)
```

```python
for ttype in ['fam', 'nov']:
    fig, ax = plt.subplots(1,3)
    fig.subplots_adjust(wspace=.5)
    fig.suptitle(ttype)
    mu_ctrl = np.array([f.mean(axis=-1) for f in ctrl_corr[ttype]]).mean(axis=0)
    mu_ko = np.array([f.mean(axis=-1) for f in ko_corr[ttype]]).mean(axis=0)

    for i in range(6):
        for j in range(6):
            if i>=j:
                mu_ctrl[i,j]=np.nan
                mu_ko[i,j]=np.nan
    # mu_ctrl, mu_ko = np.array(ctrl_corr[ttype]).mean(axis=0), np.array(ko_corr[ttype]).mean(axis=0)
    
    
    h = ax[0].imshow(mu_ctrl, vmin=0, vmax=.7, cmap='cividis')
    plt.colorbar(h, 
                shrink=.3, 
                ticks = np.arange(0,.81,.2)
                )
    ax[0].set_title('control')
    
    h = ax[1].imshow(mu_ko, vmin=0, vmax=.7, cmap='cividis')
 
    plt.colorbar(h, 
                shrink=.3, 
                ticks = np.arange(0,.81,.2)
                )
    ax[1].set_title('ko')
    
    h = ax[2].imshow(mu_ctrl-mu_ko, vmin=-.2, vmax=.2, cmap='RdGy')
    plt.colorbar(h, 
                shrink=.3, 
                ticks = [-.1, 0, .1]
                )
    ax[2].set_title('ctrl - ko')

    for a in ax.flatten():
        a.set_xticks(np.arange(6), labels = np.arange(1,7))
        a.set_yticks(np.arange(6), labels = np.arange(1,7))

        a.set_xlabel('day')
        a.set_ylabel('day')

    fig.savefig(figdir / f'cell_corr_acrossdays_{ttype}.svg')
```

```python
rows = []
for ttype in ('fam', 'nov'):
    mu_ctrl = np.array([f.mean(axis=-1) for f in ctrl_corr[ttype]])
    mu_ko = np.array([f.mean(axis=-1) for f in ko_corr[ttype]])
    # mu_ctrl, mu_ko = np.array(ctrl_corr[ttype]), np.array(ko_corr[ttype])
    
    for r in range(mu_ctrl.shape[0]):
        tmp = mu_ctrl[r,:,:]
        tmp[np.diag_indices_from(tmp)]=np.nan
        mu_ctrl[r, :, :]=tmp
    
    for r in range(mu_ko.shape[0]):
        tmp = mu_ko[r,:,:]
        tmp[np.diag_indices_from(tmp)]=np.nan
        mu_ko[r,:,:]=tmp
    
    
    mu_ctrl = np.nanmean(mu_ctrl.reshape([mu_ctrl.shape[0],-1]), axis=-1)
    mu_ko = np.nanmean(mu_ko.reshape([mu_ko.shape[0],-1]), axis=-1)

    for i, mu in enumerate(mu_ctrl):
        rows.append({
            'cond': 'ctrl',
            'ttype': ttype,
            'mouse': i,
            'pvc_mu': mu,
        })
    
    for j, mu in enumerate(mu_ko):
        rows.append({
            'cond': 'ko',
            'ttype': ttype,
            'mouse': i+j,
            'pvc_mu': mu,
        })
mu_df = pd.DataFrame(rows)


```

```python
import seaborn as sns

mu_df['tmp_x'] = 0

fig,ax = plt.subplots(1,2, figsize=[4,3], sharey = True)
for i, nov in enumerate(('fam','nov')):
    if nov == 'nov':
        marker = 'P'
        s = 8
    else:
        marker = '.'
        s = 8
    sns.stripplot(data=mu_df.loc[mu_df['ttype']==nov,:],
                x='tmp_x',
                y='pvc_mu',
                hue='cond', 
                hue_order=('ctrl', 'ko'), 
                palette=['black', 'red'],
                dodge=True,
                ax=ax[i], marker = marker, s=s,
                legend=False)
    
    sns.pointplot(data=mu_df.loc[mu_df['ttype']==nov,:],
              x='tmp_x',
              y='pvc_mu',
              errorbar=None,
              hue='cond',
              hue_order = ['ctrl', 'ko'],
              dodge = .4,
              palette = ['black', 'red'],
              ax=ax[i],
              linestyles="none",
              marker="_", markersize=15, markeredgewidth=5, legend=False, alpha=.4,
              )
    ax[i].set_title(nov)

for a in ax:
    a.spines['top'].set_visible(False)
    a.spines['right'].set_visible(False)
    a.set_ylim([0,.7])
    a.set_xticks([])

fig.savefig(figdir / f'cell_corr_acrossdays_stripplot.pdf')
```

```python

rows = []

for mouse in sparse_mice:
    print(mouse)
    if mouse == 'SparseKO_09' :#and day in set((1,2)):
            continue
    for day in range(5): 
        
        
        for next_day in range(day+1,6):
            sess = u.sparse_concat_sessions(mouse, [day, next_day], trial_mat_keys=['channel_0_F_dff', 'channel_1_F_dff'])
            rz_early = (np.argwhere(sess.trial_matrices['bin_edges'][:-1]>=sess.rzone_early['tfront'])[0], np.argwhere(sess.rzone_early['tback']<=sess.trial_matrices['bin_edges'][1:])[0] )
            rz_late = (np.argwhere(sess.trial_matrices['bin_edges'][:-1]>=sess.rzone_late['tfront'])[0], 29 )
            
            if sess.novel_arm == 1: 
                rz_nov = rz_late
                rz_fam = rz_early
            else: 
                rz_nov = rz_early
                rz_fam = rz_late

            # print(sess.trial_matrices['channel_0_F_dff'].shape, sess.trial_matrices['channel_1_F_dff'].shape)

            # pc_mask = sess.fam_place_cell_mask()*sess.nov_place_cell_mask()
            
            # # print(pc_mask.shape)

            nov_mask= sess.trial_info['LR']==sess.novel_arm
            # print("nov_mask", nov_mask.sum())
            # for day in range(5):
            for chan in ('channel_0', 'channel_1'):
                pc_mask = (sess.place_cell_info[chan][-1]['masks'].sum(axis=0)>0) * (sess.place_cell_info[chan][1]['masks'].sum(axis=0)>0)
                pc_mask = np.ones((sess.trial_matrices[f'{chan}_F_dff'].shape[-1],))>0
                # print(mouse, day, next_day, chan, pc_mask.sum())
                if pc_mask.sum()<10:
                    
                    continue
                day_mask = sess.trial_info['sess_num_ravel']==0
                # print("day_mask", day_mask.sum())

                tmat_nov = np.nanmean(sess.trial_matrices[f'{chan}_F_dff'][nov_mask*day_mask,:,:][:,:,pc_mask], axis=0)
                

                tmat_fam = np.nanmean(sess.trial_matrices[f'{chan}_F_dff'][~nov_mask*day_mask,:,:][:,:,pc_mask], axis=0)
                
                next_day_mask = sess.trial_info['sess_num_ravel']==1
                # print("next_day_mask", next_day_mask.sum())
                for ttype in ('nov', 'fam'):
                    if ttype == 'nov':
                        n_mask = nov_mask
                        day_mat = tmat_nov          
                        
                    else:
                        n_mask = ~nov_mask
                        day_mat = tmat_fam

                        
                    # print((n_mask*next_day_mask).sum(), pc_mask.sum())
                    next_day_tmat = np.nanmean(sess.trial_matrices[f'{chan}_F_dff'][n_mask*next_day_mask,:,:][:,:,pc_mask],axis=0)
                    
                    
                    
                    next_day_corr = np.zeros([next_day_tmat.shape[1],])
                    for cell in range(next_day_tmat.shape[1]):
                        r, _ = sp.stats.pearsonr(day_mat[:,cell], next_day_tmat[:,cell])
                        next_day_corr[cell]=r

                    # print(chan)

                    for ndc in next_day_corr:
                        rows.append({'chan': chan,
                                    'mouse': mouse, 
                                    'day': day,
                                    'next_day': next_day,
                                    'ttype': ttype,
                                    'next_day_corr': ndc,
                                    })
                            

df = pd.DataFrame(rows)
```

```python
# correlation matrices across days

corr_mat = {}
for chan in ('channel_0','channel_1'):
    corr_mat[chan] = {'fam': None,
                      'nov': None,
                     }
    print(chan)

    for ttype in ('fam', 'nov'):
        
        mat = np.nan*np.zeros([6,6, len(sparse_mice)])
        for day in range(5):
            for next_day in range(day+1, 6):
            
                mask = (df['chan']==chan) * (df['day']==day) * (df['next_day']==next_day) * (df['ttype']==ttype) 
                tmp_df = df.loc[mask].groupby(['mouse'])['next_day_corr'].mean().reset_index()
                # print(tmp_df['mouse'].unique())
                # print(cond, ttype, reward_cell, day, next_day)
                for i, mouse in enumerate(sparse_mice):
                    mouse_mask = tmp_df['mouse']==mouse
                    if mouse_mask.sum()>0:
                        mat[day, next_day,i] = tmp_df.loc[mouse_mask,'next_day_corr'].values[0]

            
        corr_mat[chan][ttype]= mat
          


```

```python
fig, ax = plt.subplots(2,3)
fig.subplots_adjust(wspace=.5, hspace=.5)

for i, ttype in enumerate(('fam', 'nov')):
    for j, chan in enumerate(('channel_1', 'channel_0')):
        mat = corr_mat[chan][ttype]
        h = ax[i,j].imshow(np.nanmean(mat,axis=-1), cmap='cividis', vmin=0, vmax=.7)
        ax[i,j].set_title(f'{chan} : {ttype}')
        fig.colorbar(h, shrink=.6)

    mat = corr_mat['channel_1'][ttype] - corr_mat['channel_0'][ttype]
    h = ax[i,2].imshow(np.nanmean(mat, axis=-1), cmap='RdGy', vmin=-.2, vmax=.2)
    ax[i,2].set_title('channel_1 - channel_0')
    fig.colorbar(h, shrink=.6)

for a in ax.flatten():
    a.set_xticks(np.arange(6), labels = np.arange(1,7))
    a.set_yticks(np.arange(6), labels = np.arange(1,7))

    a.set_xlabel('day')
    a.set_ylabel('day')
    
fig.savefig(figdir / f'chan_comparison_next_day_corr_heatmaps.svg')
```
