```python
import pathlib
import numpy as np
import scipy as sp
from matplotlib import pyplot as plt


import STX3KO_analyses as stx
from STX3KO_analyses import utilities as u

ko_mice = stx.ymaze_sess_deets.ko_mice
ctrl_mice = stx.ymaze_sess_deets.ctrl_mice
sparse_mice = stx.ymaze_sess_deets.sparse_mice

plt.rcParams['pdf.fonttype']=42
plt.rcParams['font.family'] ='sans-serif'
plt.rcParams['font.serif'] = 'Arial'


%matplotlib inline

%load_ext autoreload
%autoreload 2
```

```python
figdir = pathlib.Path('/mnt/BigDisk/fig_scratch/stx3/fig2_normal_placecells')
```

```python
# pv correlation across days
tmat_key = 'F_dff'
ctrl_corr = {'fam': [], 'nov': []}
for mouse in ctrl_mice:
    pv_corr = stx.spatial_analyses.run_acrossday_corr(mouse, fam=True, key=tmat_key, method='pv')
    ctrl_corr['fam'].append(pv_corr)

    pv_corr = stx.spatial_analyses.run_acrossday_corr(mouse, fam=False, key=tmat_key, method='pv')
    ctrl_corr['nov'].append(pv_corr)


ko_corr = {'fam': [], 'nov':[]}
for mouse in ko_mice:
    pv_corr = stx.spatial_analyses.run_acrossday_corr(mouse, fam=True, key=tmat_key, method='pv')
    ko_corr['fam'].append(pv_corr)

    pv_corr = stx.spatial_analyses.run_acrossday_corr(mouse, fam=False, key=tmat_key, method='pv')
    ko_corr['nov'].append(pv_corr)
```

```python
for ttype in ['fam', 'nov']:
    fig, ax = plt.subplots(1,3)
    fig.subplots_adjust(wspace=.5)
    fig.suptitle(ttype)
    mu_ctrl, mu_ko = np.array(ctrl_corr[ttype]).mean(axis=0), np.array(ko_corr[ttype]).mean(axis=0)
    
    h = ax[0].imshow(mu_ctrl, vmin=0, vmax=.8, cmap='cividis')
    plt.colorbar(h, 
                shrink=.3, 
                ticks = np.arange(0,.81,.2)
                )
    ax[0].set_title('control')
    
    h = ax[1].imshow(mu_ko, vmin=0, vmax=.8, cmap='cividis')
 
    plt.colorbar(h, 
                shrink=.3, 
                ticks = np.arange(0,.81,.2)
                )
    ax[1].set_title('ko')
    
    h = ax[2].imshow(mu_ctrl-mu_ko, vmin=-.2, vmax=.2, cmap='RdGy')
    plt.colorbar(h, 
                shrink=.3, 
                ticks = [-.1, 0, .1]
                )
    ax[2].set_title('ctrl - ko')

    for a in ax.flatten():
        a.set_xticks(np.arange(6), labels = np.arange(1,7))
        a.set_yticks(np.arange(6), labels = np.arange(1,7))

        a.set_xlabel('day')
        a.set_ylabel('day')

    fig.savefig(figdir / f'pv_corr_acrossdays_{ttype}.svg')
```
