```python
# check this
def dense_get_pc_mask(sess):
    try:
        return {
            -1: sess.place_cell_info['left'],
            1: sess.place_cell_info['right'],
            }
    except:
        return {
            -1: sess.place_cell_info[-1],
            1: sess.place_cell_info[1],
            }

def sparse_get_pc_mask(sess):
    return {
        'channel_0': {
            -1: sess.place_cell_info['channel_0_F_dff']['left'],
            1: sess.place_cell_info['channel_0_F_dff']['right'],
        },
        'channel_1': {
            -1: sess.place_cell_info['channel_1_F_dff']['left'],
            1: sess.place_cell_info['channel_1_F_dff']['right'],
        }
    }
    
```

```python
import pickle
for old_name, new_name in mouse_lookup.items():
    
    for day in range(6):
        
        # load single day 
        
        if old_name == "SparseKO_09" and day==2:
            continue

        out_dir = pathlib.Path(f'/home/mplitt/repos')
        out_dir = out_dir / 'Plitt_Kaganovsky_Stx3_CA1' / 'data' / 'place_cell_masks' / new_name / f'day{day}'
        out_dir.mkdir(exist_ok=True, parents=True)

        # if (out_dir / 'pc_mask.pkl').is_file():
        #     continue


        sess = stx.utilities.load_single_day(old_name,day)
        
        if 'Sparse' not in old_name:
            pc_dict = dense_get_pc_mask(sess)
        else:
            pc_dict = sparse_get_pc_mask(sess)


        with open(str(out_dir / 'pc_mask.pkl'),'wb') as file:
            pickle.dump(pc_dict, file)
```
