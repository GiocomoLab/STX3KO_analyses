```python
import STX3KO_analyses as stx
import pathlib
import subprocess
```

```python
with open('.oak_pwd', 'r') as f:
        oak_pwd = f.read().strip()
```

```python
sparse_mice = list(stx.mouse_metadata.sparse_sessions.keys())

```

```python
mouse = sparse_mice[0]
metadata = stx.mouse_metadata.sparse_sessions.get(mouse)
sessions = metadata.get('sessions')
```

```python
def sbx_rsync(mouse, session):
    remote_user = "mplitt"
    remote_host = "dtn.sherlock.stanford.edu"

    if mouse == 'SparseKO_02':
        remote_base_path = pathlib.Path("/oak/stanford/groups/giocomo/InVivoDA/2P_Data/")
    else:
        remote_base_path = pathlib.Path("/oak/stanford/groups/giocomo/mplitt/2P_Data/STX3KO/")
    
    session_dir = remote_base_path / mouse / session.get("date_str") / session.get("scene")
    sbx_mat_filename = f"{session.get('scene')}_{session.get('session'):03}_{session.get('scan'):03}.mat"
    sbx_mat_path = str(session_dir / sbx_mat_filename)
    print(sbx_mat_path)

    local_path = pathlib.Path('/mnt/BigDisk/2P_scratch')
    local_path = local_path / mouse / session.get("date_str") / session.get("scene")
    local_path.mkdir(exist_ok=True, parents=True)

    cmd = [
        "sshpass", "-p", oak_pwd,
        "rsync", "-rlt", "--progress", 
        f"{remote_user}@{remote_host}:{sbx_mat_path}",
        str(local_path) 
    ]
    # print(cmd)
    
    try:
        subprocess.run(cmd, check=True)
        print("Rsync completed successfully.")
    except subprocess.CalledProcessError as e:
        print(f"Rsync failed with error: {e}")
        
   
```

```python
for mouse, metadata in stx.mouse_metadata.sparse_sessions.items():
    for i, session in enumerate(metadata.get('sessions')):
        if session is None:
            continue
        # print(session)
        print(mouse, i)
        sbx_rsync(mouse, session)
```
