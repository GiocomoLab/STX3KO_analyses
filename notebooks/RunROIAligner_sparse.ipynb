```python
import os
import dill
from matplotlib import pyplot as plt
import numpy as np

import TwoPUtils
from TwoPUtils.roi_matching import ROIAligner

import STX3KO_analyses as stx

%matplotlib inline

%load_ext autoreload
%autoreload 2
```

```python
mouse = 'SparseKO_06'
sessions = stx.ymaze_sess_deets.SparseKO_sessions[mouse]
sess_dir = os.path.join('/home/mplitt/YMazeSessPkls',mouse)
sess_list = []
for sess_deets in sessions:     
    if isinstance(sess_deets,(list,tuple)):
        for _sess_deets in sess_deets:
            with open(os.path.join(sess_dir,_sess_deets['date'],"%s_%d.pkl" % (_sess_deets['scene'], _sess_deets['session'])),'rb') as file:
                sess_list.append(dill.load(file))
    else:
        with open(os.path.join(sess_dir,sess_deets['date'],"%s_%d.pkl" % (sess_deets['scene'], sess_deets['session'])),'rb') as file:
                sess_list.append(dill.load(file))

for sess in sess_list:
    for channel in ('channel_0','channel_1'):

        if sess.s2p_ops[channel]['meanImg'].shape[1]==796:
            sess.s2p_ops[channel]['meanImg'] = sess.s2p_ops[channel]['meanImg'][:,80:746]
            sess.s2p_ops[channel]['Lx'] = sess.s2p_ops[channel]['Lx'] - 80-50
            # sess.s2p_ops[channel]['meanImg'][:,-50:] = 0

            for cell in sess.s2p_stats[channel]:
                cell['xpix'] = cell['xpix'] - 80
                # cell['xpix'][cell['xpix']>646] = 646


        fig, ax = plt.subplots(1,1)
        img = sess.s2p_ops[channel]['meanImg']
        ax.imshow(sess.s2p_ops[channel]['meanImg'], cmap='gray')
        for cell in sess.s2p_stats[channel]:
            y,x = cell['ypix'], cell['xpix']
            ax.plot(x.mean(), y.mean(),'r.')
```

```python

for chan in ('channel_0','channel_1'):
    aligner = ROIAligner(sess_list, channel=chan)
    aligner.run_pairwise_matches()
    
    with open(os.path.join(sess_dir,f'roi_aligner_results_{chan}.pkl'), 'wb') as file:
            dill.dump(aligner.match_inds,file)
```

```python
mouse = 'SparseKO_02'
sessions = stx.ymaze_sess_deets.SparseKO_sessions[mouse]
sess_dir = os.path.join('/home/mplitt/YMazeSessPkls',mouse)
sess_list = []
for sess_deets in sessions:     
    if isinstance(sess_deets,(list,tuple)):
        for _sess_deets in sess_deets:
            with open(os.path.join(sess_dir,_sess_deets['date'],"%s_%d.pkl" % (_sess_deets['scene'], _sess_deets['session'])),'rb') as file:
                sess_list.append(dill.load(file))
    else:
        with open(os.path.join(sess_dir,sess_deets['date'],"%s_%d.pkl" % (sess_deets['scene'], sess_deets['session'])),'rb') as file:
                sess_list.append(dill.load(file))

for sess in sess_list:
    for channel in ('channel_0','channel_1'):
        img = sess.s2p_ops[channel]['meanImg']
        

        if sess.s2p_ops[channel]['meanImg'].shape[1]==796:
            sess.s2p_ops[channel]['meanImg'] = sess.s2p_ops[channel]['meanImg'][:,100:746]
            sess.s2p_ops[channel]['Lx'] = sess.s2p_ops[channel]['Lx'] - 100 -50
            # sess.s2p_ops[channel]['meanImg'][:,-50:] = 0

            for cell in sess.s2p_stats[channel]:
                cell['xpix'] = cell['xpix'] - 100
                # cell['xpix'][cell['xpix']>646] = 646


        elif sess.s2p_ops[channel]['meanImg'].shape[1]==786:
            sess.s2p_ops[channel]['meanImg'] = sess.s2p_ops[channel]['meanImg'][:,100:746]
            sess.s2p_ops[channel]['Lx'] = sess.s2p_ops[channel]['Lx'] - 100 -40
            # sess.s2p_ops[channel]['meanImg'][:,-50:] = 0

            for cell in sess.s2p_stats[channel]:
                cell['xpix'] = cell['xpix'] - 100
                # cell['xpix'][cell['xpix']>646] = 646
        elif sess.s2p_ops[channel]['meanImg'].shape[1]==696:
            sess.s2p_ops[channel]['meanImg'] = sess.s2p_ops[channel]['meanImg'][:,:646]
            sess.s2p_ops[channel]['Lx'] = sess.s2p_ops[channel]['Lx'] - 50             

            for cell in sess.s2p_stats[channel]:
                cell['xpix'] = cell['xpix'] 
             
       
        if sess.s2p_ops[channel]['meanImg'].shape[0]<512:
            img = np.copy(sess.s2p_ops[channel]['meanImg'])
            sess.s2p_ops[channel]['meanImg'] = np.pad(img,((512 - img.shape[0],0),(0,0)),'constant', constant_values=np.amin(img))
            sess.s2p_ops[channel]['Ly'] = 512

            for cell in sess.s2p_stats[channel]:
                cell['ypix'] = cell['ypix'] + (512 - img.shape[0])

        for cell in sess.s2p_stats[channel]:
            y,x = cell['ypix'], cell['xpix']
            mask = (y<512) * (x<646)
            cell['ypix'] = y[mask]
            cell['xpix'] = x[mask]
        fig, ax = plt.subplots(1,1)
        img = sess.s2p_ops[channel]['meanImg']
        print(img.shape)
        ax.imshow(img, cmap='gray', vmin=np.percentile(img,5), vmax=np.percentile(img,99))
        for cell in sess.s2p_stats[channel]:
            y,x = cell['ypix'], cell['xpix']
            ax.plot(x.mean(), y.mean(),'r.')
```

```python

for chan in ('channel_0','channel_1'):
    aligner = ROIAligner(sess_list, channel=chan)
    aligner.run_pairwise_matches()
    
    with open(os.path.join(sess_dir,f'roi_aligner_results_{chan}.pkl'), 'wb') as file:
            dill.dump(aligner.match_inds,file)
```

```python
def single_mouse_aligner(mouse, sessions):
    sess_dir = os.path.join('/home/mplitt/YMazeSessPkls',mouse)
    sess_list = []
    for sess_deets in sessions:     
        if isinstance(sess_deets,(list,tuple)):
            for _sess_deets in sess_deets:
                with open(os.path.join(sess_dir,_sess_deets['date'],"%s_%d.pkl" % (_sess_deets['scene'], _sess_deets['session'])),'rb') as file:
                    sess_list.append(dill.load(file))
        else:
            with open(os.path.join(sess_dir,sess_deets['date'],"%s_%d.pkl" % (sess_deets['scene'], sess_deets['session'])),'rb') as file:
                    sess_list.append(dill.load(file))

    for chan in ['channel_0','channel_1']:
        sa = ROIAligner(sess_list, channel=chan)
        sa.run_pairwise_matches()

        with open(os.path.join(sess_dir,f'roi_aligner_results_{chan}.pkl'), 'wb') as file:
            dill.dump(sa.match_inds,file)
    
```
