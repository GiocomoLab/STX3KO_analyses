```python
def quadrant_mask(Ly,Lx,ny,nx,sT):
    mask = np.zeros((Ly,Lx), np.float32)
    mask[np.ix_(ny,nx)] = 1
    mask = sp.ndimage.gaussian_filter(mask, sT)
    return mask

def correct_bleedthrough(Ly, Lx, nblks, mimg, mimg2):
    # subtract bleedthrough of green into red channel
    # non-rigid regression with nblks x nblks pieces
    sT = np.round((Ly + Lx) / (nblks*2) * 0.25)
    mask = np.zeros((Ly, Lx, nblks, nblks), np.float32)
    weights = np.zeros((nblks, nblks), np.float32)
    yb = np.linspace(0, Ly, nblks+1).astype(int)
    xb = np.linspace(0, Lx, nblks+1).astype(int)
    for iy in range(nblks):
        for ix in range(nblks):
            ny = np.arange(yb[iy], yb[iy+1]).astype(int)
            nx = np.arange(xb[ix], xb[ix+1]).astype(int)
            mask[:,:,iy,ix] = quadrant_mask(Ly, Lx, ny, nx, sT)
            x  = mimg[np.ix_(ny,nx)].flatten()
            x2  = mimg2[np.ix_(ny,nx)].flatten()
            # predict chan2 from chan1
            A = np.ones([x.shape[0],2])
            A[:,0] = x
            a,_ = sp.optimize.nnls(A,x2)
#             print(a)
#             a = (x * x2).sum() / (x * x).sum()
            weights[iy,ix] = a[0]
    mask /= mask.sum(axis=-1).sum(axis=-1)[:,:,np.newaxis,np.newaxis]
    mask *= weights
    mask *= mimg[:,:,np.newaxis,np.newaxis]
    print((mask.sum(axis=-1).sum(axis=-1)<0).sum())
    mimg2 -= mask.sum(axis=-1).sum(axis=-1)
    mimg2 = np.maximum(0, mimg2)
    return mimg2
```

```python

for file in file_list:
    if file['scene'] in ('combined'):
        s2p_folder = os.path.join(basedir, mouse, file['date'],'combined','suite2p','plane0')
    else:
        s2p_folder = os.path.join(basedir, mouse, file['date'],file['scene'], "%s_%03d_%03d" % (file['scene'], file['session'], file['scan']), 'suite2p','plane0')
        
    ops = np.load(os.path.join(s2p_folder,"ops.npy"),allow_pickle=True).all()
    stats = np.load(os.path.join(s2p_folder,"stat.npy"),allow_pickle=True)
    iscell = np.load(os.path.join(s2p_folder,"iscell.npy"))
#     iscell[:,0]=1
    
    
    mimg2 = correct_bleedthrough(ops['Ly'], ops['Lx'], 8, ops['meanImg'], ops['meanImg_chan2'])
    
    bckgnd = sp.ndimage.filters.gaussian_filter(mimg2,20)
    
    mimg2_corr = mimg2/bckgnd

    fig,ax = plt.subplots(1,4,figsize=[30,10])
    ax[0].imshow(mimg2)
    ax[1].imshow(bckgnd)
    ax[2].imshow(mimg2_corr)
    
    
    mask = np.ones(mimg2.shape)
    mask[mimg2_corr<.9] = 0
    ax[3].imshow(mimg2_corr*mask)
    fig,ax = plt.subplots(figsize=[30,30])
    ax.imshow(mimg2_corr*mask, vmin = .2, vmax = 2)

    for i, roi in enumerate(stats):
        if mask[roi['ypix'],roi['xpix']].sum()<.5*roi['ypix'].shape[0]:
            iscell[i,0] = 0

            
    np.save(os.path.join(s2p_folder,"iscell.npy"),iscell)
        
    
```
